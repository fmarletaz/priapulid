```python
from glob import glob
from collections import defaultdict
from copy import deepcopy
from tqdm.notebook import trange, tqdm
```

```python
dupsCt=defaultdict(int)
lossCt=defaultdict(int)
outL=[]
outD=[]
for f in tqdm(glob("Priap_0124_xml/*.xml")):
    RTL = parser.parse(f)
    Evs=RTL.getEventsSummary()
    dupsNd=defaultdict(int)
    lossNd=defaultdict(int)
    for n in Evs['duplication']:
        dupsCt[n]+=1
        dupsNd[n]+=1
    for n in Evs['loss']:
        lossCt[n]+=1
        lossNd[n]+=1
    l=[f]+[lossNd[n] for n in spNodes]
    d=[f]+[dupsNd[n] for n in spNodes]
    outL.append(l)
    outD.append(d)
    

print(dupsCt)
print(lossCt)
```

```python
spTd=deepcopy(RTL.spTree)
for l in spTd.traverse("postorder"):
    nm=l.name
    l.dist="{}".format(dupsCt[nm]/sum(dupsCt.values()))
    if not l.is_leaf():
        l.name="{}".format(dupsCt[nm])
```

```python
print(spTd.get_ascii(show_internal=True))
```

```python
spTl=deepcopy(RTL.spTree)
for l in spTl.traverse("postorder"):
    nm=l.name
    l.dist="{}".format(lossCt[nm]/sum(lossCt.values()))
    #l.name="{}:{}/{}".format(nm,dupsCt[nm],lossCt[nm])
    if not l.is_leaf():
        l.name="{}".format(lossCt[nm])
```

```python
print(spTl.get_ascii(show_internal=True))
```
